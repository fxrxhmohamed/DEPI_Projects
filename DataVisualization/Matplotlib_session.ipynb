# This is Data Visualization session using Matplotlib

# Line Plot

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
amounts = {
    'Black': [5, 7, 8, 6],
    'White': [6, 9, 7, 8]
}
width = 0.5

fig, ax = plt.subplots()
bottom = np.zeros(len(categories))
# Stacker bar plot
for color, amount in amounts.items():
    ax.bar(categories, amount, width, bottom=bottom, label=color)
    bottom += amount

ax.set_title("Number of penguins with above average body mass")
ax.legend(loc="upper right")

plt.show()
```
