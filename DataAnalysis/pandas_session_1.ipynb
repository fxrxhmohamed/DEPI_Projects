# Pandas for Data Analysis 

# Create DataFrames

```python
import pandas as pd

data = pd.DataFrame({
    'IDs': [1,2,3,4,5],
    'Names': ['farah', 'moh', 'abd', 'salma', 'doaa'],
    'Ages': [24,32,19,20,21]
})

print(data)
```

```python
# DataFrame Overview
# info() -> print all columns, there count, types , whether there is nulls
print(data.info())  
#describe() -> statistical info
print(data.describe())
```

# Reshaping data
* melt()
* pivot()
* concat()
* concat() along columns

```python
# melt() -> btsaya7 el columns, gather columns into rows

data3 = pd.DataFrame(
    {
        'Names' : ['alice', 'bob', 'charlie'],
        'Cairo' : [1, 0, 1],
        'Alex' : [0, 1, 0]
    }
)

data_melted = pd.melt(data3, id_vars=["Names"], var_name='City', value_name='Visited')
print(data_melted)
```

```python
# pivot() -> byrag3 el melt lel original, spread rows into columns

data_pivot = data_melted.pivot(index='Names', columns='City', values='Visited')
data_pivot['Names'] = data_pivot.index
data_pivot = data_pivot.reset_index(drop=True) # 34n el Names matkonsh el index htkon column, el index htkon arqam 3adi [0,1,2]
print(data_pivot)
```

```python
# concat()
data2= pd.DataFrame({
    'IDs': [6, 7],
    'Names': ['frank', 'grace'],
    'Ages': [20,30]
})

data = pd.concat([data, data2], ignore_index=True)
print(data)
```

```python
# # concatenate along columns

# Method 1
data['City'] = ['Cairo', 'Alex', 'Cairo', 'Alex', 'Cairo', 'Alex', 'Cairo']
print(data)

# # Method 2
# # axis=1 means add to columns, axis=0 by default and means add to rows
# data = pd.concat([data, data2], axis=1) 
# print(data)
```

```python
# Sorting
data = data.sort_values(by='Ages', ascending=False) # By default ascending=True
print(data)
print('=============')

# Renaming  -> rename column
data = data.rename(columns={'IDs':'ID', 'Names':'Name', 'Ages':'Age', 'City':'Location'})
print(data)
print('=============')

# drop columns
data = data.drop(columns=['ID'])
print(data)
print('=============')

# drop rows 
data = data.drop(index=[0, 1])
print(data)
```

# Selection from dataframe

```python
data = pd.DataFrame({
    'ID': [1,2,3,4,5, 6, 7],
    'Name': ['farah', 'moh', 'abd', 'salma', 'doaa', 'frank', 'grace'],
    'Age': [24,32,19,20,21, 20, 30],
    'City': ['Cairo', 'Alex', 'Cairo', 'Alex', 'Cairo', 'Alex', 'Cairo']
})

print(data)
```

```python
# select based on a condition
data[data['Age']>21]
```

```python
# select certain rows -> 'iloc' means based on the 'index location'
data.iloc[:3]
```

```python
# select based on multiple criterias
data[(data['Age'] > 21) & (data['City'] == 'Cairo')]
```
