# XPath
* What is XPath?
-> a way to search for element similar to RegEx form, XPath lets you select nodes/elements in that tree using a path-like syntax, Can target elements using tags, attributes, text, or hierarchy.
* Why is XPath used in web scraping?
-> Sometimes web pages don’t have convenient IDs or class names, so traditional selectors like By.ID or By.CLASS_NAME may not work.

```python
# get all elements
from selenium.webdriver import Chrome
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

chrome_options = Options()
chrome_options.add_argument("--headless")  # Run in headless mode

driver = Chrome(
    options=chrome_options
)

driver.get("https://www.w3schools.com/sql/")

# # find element using tag name
# a_elements = driver.find_elements(By.TAG_NAME, "a")
# for a in a_elements:
#     print(a.get_attribute('href'))


# find element using xpath
a_elements = driver.find_elements(By.XPATH, "//div/div[1]/h2")
for a in a_elements:
    print(a.text)

driver.quit()
```

# Note: 
Use implicitly_wait() better than sleep() as it is driver based so more focused and robust

# this code does:
 1) Finds the search box, types "Python", and presses ENTER.
 2)Finds the footer element and scrolls to it using ActionChains.

```python
# get all elements
from selenium.webdriver import Chrome
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver import ActionChains  # ActionChains → Allows you to perform advanced interactions like hovering, dragging, or scrolling.

chrome_options = Options()

driver = Chrome(
    options=chrome_options
)

driver.get("https://www.w3schools.com/sql/")

driver.implicitly_wait(2)

search_element = driver.find_element(By.XPATH, "/html/body/div[2]/div[1]/div[1]/div[1]/input")
if search_element:
    search_element.send_keys("Python")
    search_element.send_keys(Keys.ENTER)

footer = driver.find_element(By.XPATH, '//*[@id="footerwrapper"]')
ActionChains(driver)\
    .move_to_element(footer)\
    .perform()

```
