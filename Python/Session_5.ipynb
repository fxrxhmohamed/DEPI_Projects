# Constructor creatin example: Calculator 

```python
class calculator:
    def __init__(self,x,y):
        print("A calculator have been created!")
        self.x=x
        self.y=y
    def add(self):
        return self.x + self.y
    
    def substract(self):
        return self.x - self.y
    
    def multiply(self):
        return self.x * self.y

    def divide(self):
        return self.x / self.y

c = calculator(4,2)
print(c.add())
print(c.substract())
print(c.multiply())
print(c.divide())

    
```

# Inheritance

```python
class pet():
    def __init__(self,name,color,voice):
        self.name=name
        self.color=color
        self.voice=voice
    def speak(self):
        print(self.voice,"...",self.voice)

class cat(pet):
    def __init__(self,name,color):
        super().__init__(name,color,"Meow")

class dog(pet):
    def __init__(self,name,color):
        super().__init__(name,color,"Woof")

tom = cat("Tom","Grey")
tom.speak()
roy=dog("Roy","Black")
roy.speak()
```

# Task 
Add to the Country class so that it contains the attribute is_big. Set it to True if either criterion are met:
Population is greater than 250 million.
Area is larger than 3 million square km.
Also, create a method which compares a country’s population density to another country object

```python
class Country():
    def __init__(self,name,population,area):
        self.name = name
        self.population=population
        self.area=area

    def is_big(self):
        if self.population > 250000000 or self.area > 3000000:
            return True
        else:
            return False
        
    def compare_pd(self,other):
        self_density = self.population / self.area
        other_density = other.population / other.area

        if self_density > other_density:
            return f"{self.name} has a larger population density than {other.name}"

        elif self_density < other_density:
            return f"{other.name} has a larger population density than {self.name}"
        else:
            return f"{other.name} has a larger population density than {self.name}"
        
australia = Country("Australia", 260000000,30)
print(australia.is_big())

egypt = Country("Egypt",250,35000000)
print(egypt.is_big())

turkie = Country("Turkie",250,35000000)
print(turkie.is_big())

japan = Country("Japan",250,30)
print(japan.is_big())

print(japan.compare_pd(egypt))
print(turkie.compare_pd(egypt))
```

```python
# each attribute/method in a class have 3 levels of abstraction (access modifiers)
# 1) public: "don't have any underscores before attribute name" normal attributes can be accessed from inside/outside the class and can be inherited
# 2) protected: "will has 1 underscore before attribute name" attribute/method that should not be accessed outside the class visible 
# 3) private: "will have 2 underscores before the attribute" attribute/method can't be access from outside the class and class not visible 
# Q) what is the use of the private modifier? ->  restrict access to class attributes or methods from outside the class.which helps keep internal class details hidden and protected. u can access this attributes using setters and getters

#Solution 1

class pet():
    def __init__(self,name,color,voice):
        self.name=name #publiv attribute
        self.color=color
        self.voice=voice

        self._A1 = "Protected"
        self.__A2 = "Private"
    def speak(self):
        print(self.voice,"...",self.voice)

class cat(pet):
    def __init__(self,name,color):
        super().__init__(name,color,"Meow")

class dog(pet):
    def __init__(self,name,color):
        super().__init__(name,color,"Woof")

    
tom = cat("Tom","Grey")
tom.speak()
roy=dog("Roy","Black")
roy.speak()

print(tom._A1)
# print(tom.__A2)  # will raise this error "cat' object has no attribute '__A2'"  as __A2 is private can't be accessed without setters & getters

# in the coming code will resolve this problem using setters and getters....
```

# Abstraction example

```python
from abc import ABC, abstractmethod

#Abstract classes provide a blueprint for other classes.
#They cannot be instantiated directly and require subclasses to implement abstract methods.
class Vehicle(ABC):
   @abstractmethod
   def start_engine(self):
       pass

class Car(Vehicle):
    def start_engine(self):
       print("Car engine started.")

class Bike(Vehicle):
     def start_engine(self):
        print("Bike engine started.")

#The following line would raise an error:
#v = Vehicle() # TypeError: Can't instantiate abstract class Vehicle with abstract method start_engine
car = Car()
bike = Bike()
car.start_engine()
bike.start_engine()
```

# Polymorphism

```python
# polymorphism
# is all about using the same interface for different data types.
# It allows us to define a single interface and have multiple implementations of that interface.

from abc import ABC, abstractmethod

class Shape:
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self): #this is called polymprphism as circle implements the area method defined in shape
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

shapes = [Circle(5), Square(4)]
for shape in shapes:
    print(shape.area())
# calls the area method of each shape, demonstrating polymorphism
```
