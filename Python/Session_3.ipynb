-> break:  break btksar el loop w msh btrg3 el loop tani msh hatshof el ba2e

```python
# break btksar el loop w msh btrg3 el loop tani msh hatshof el ba2e
x = 20
while x > 0:
    print(x)
    x -= 1
    if x == 10:
        break 
```

```python
# continue will skip the rest of the code in the loop and continue to the next iteration 
x = 20
while x > 0:
    if x == 10:
        x -= 1
        continue  # will only skip the 10 iteration
    print(x)
    x -= 1
    
```

# Task: 
takes a string and calculates the number of letters and digits within it.
Return the result in a dictionary.

```python
text = input("Enter a text of letters and digits")
count_letters = 0
count_digits = 0

for i in text: 
    if i.isalpha():         # check if it's a letter
        count_letters += 1
    elif i.isdigit():       # check if it's a digit
        count_digits += 1

res = {"Letters":count_letters,"Digits":count_digits}
print(res)
```

# Task:
create a function that take a list and finds the integer which appears an odd no. of times

```python
def find_odd(lst):
    lst2=[]
    for i in lst:
        if lst.count(i) % 2 != 0 and i not in lst2: # i not in lst2 ensures we don't add duplicates to lst2
            lst2.append(i)
    return lst2 # Without return, the function doesn’t send any result back—it just runs and exits, and by default returns None. So calling print(find_odd(lst)) would print None

lst = [1,1,2,-2,5,2,4,4,-1,-2,5,7,7,7] # output: -1 , 7 -> as it appears only 1 time and 3 times respectively
print(find_odd(lst))  
```

```python
# This will only print the first return value, which is x + y. The second return statement will never be executed.
def calc(x,y):
    return x + y
    return x - y

print(calc(5,3)) 

# to return multiple objects in a function: Method 1 will return a tuple
def calc2(x, y):
    return x + y, x - y

print(calc2(5, 3))  # This will print a tuple with both results

# to return multiple objects in a function: Method 2 using unpacking and function calc2
add,sub = calc2(3,9)
print(add,sub) 
```

# Arguments

# 1 positional arguments -> are always required or will give error

```python
def add(x,y):
    return x + y

print(add(5,3))  #if i did not add x,y values will give this error: TypeError: add() missing 2 required positional arguments: 'x' and 'y'
```

# 2 default arguments -> are not required, optional

```python
def add(x,y=5): 
    return x + y
print(add(3,8))
print(add(3))  # This will use the default value for y, which is 5
```

# 3 variable length argument

```python
def sum_integers(*args):
    total = 0
    for num in args:
        if isinstance(num, int):  # Check if the argument is an integer
            total += num
    return total
print(sum_integers(1, 2, '4', 3, 6, 5.0))  
```

# 4 keyword arguments


```python

def add(**kwargs):
    print(kwargs)  # This will print a dictionary of keyword arguments.

print(add(name="John", age=30, city="New York"))  # Output: {'name': 'John', 'age': 30, 'city': 'New York'}
```
