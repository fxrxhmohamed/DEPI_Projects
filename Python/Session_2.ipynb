# List -> Use Square Brackets

```python

tmp = [3,4,6,7,3,7,87,34,57,89,34, "Hello"]
print(tmp)
print(type(tmp))

# Slicing
print(tmp[3])
print(tmp[3:9])
print(tmp[-1])  
print(tmp[-2])

#modify elements
tmp[-2] = 70
print(tmp)

# Note: can't call index out of range

#Adding elements
tmp.append(100) #-> will add at the end
print(tmp)

tmp.insert(2, 200) #-> will add at index 2 value 200 w y3ml shift lel elements b3dha
print(tmp)

# Extending the list
tmp_2 = [1, 2, 3]
tmp.extend(tmp_2)  #-> will add all elements of tmp_2 to the end of tmp
print(tmp)

```

```python
tmp = [3,4,6,7,3,7,87,34]

# Removing elements
tmp.remove(7)  # Removes the first occurrence of 7
print(tmp)

tmp.pop()  # Removes the last element
print(tmp)

tmp.pop(2)  # Removes the element at index 2
print(tmp)

# Reversing the list
tmp.reverse()
print(tmp)

#sorting the list
tmp.sort()  # Sorts the list in ascending order
print(tmp)

tmp.sort(reverse = True) # Sorts the list in descending order
print(tmp)

# Sorting a list of strings alphabetically
tmp2 =["F", "K", "H", "S", "A"]
tmp2.sort()
print(tmp2)

# Sorting a list with mixed types will raise an error
# tmp2 =["F", "K", "H", 1, 4, 2]
# tmp2.sort()
# print(tmp2)


print(3 in tmp) # check if 3 exists in tmp list and print true or false
print(len(tmp)) # length of the list
print(tmp.index(3)) # Returns the index of the first occurrence of 3
print(tmp.count(3)) # Returns the count of occurrences of 3 in the list

```

```python
#Shallow copy 
a = [1,2,3]
b = a # b is a reference to the same list as a
b.append(4)  # Modifying b will also modify a
print(a)  # Output: [1, 2, 3, 4]
print(b)  # Output: [1, 2, 3, 4]

# Deep copy
a = [1, 2, 3]
b = a.copy()  # b is now a separate copy , it created another place in memory with another address so now they both do not point to the same address
b.append(4)
print(a)  # Output: [1, 2, 3]
print(b)  # Output: [1, 2, 3, 4]
```

# TASK: Given a number, return a list containing the two halves of the number. If the number is odd, make the rightmost number higher.

Examples
number_split(4) ➞ [2, 2]

number_split(10) ➞ [5, 5]

number_split(11) ➞ [5, 6]

number_split(-9) ➞ [-5, -4]
Notes
All numbers will be integers.
You can expect negative numbers too.

```python
#Solution 1
x = int(input("Enter a number: "))
if x % 2 == 0:
    split = x // 2
    print(split,split)
elif x % 2 != 0:
    split = x // 2
    print([split, split + 1])
```

```python
#Solution 2
from math import ceil
x = int(input("Enter a number: "))
if x % 2 == 0:
    split = x // 2
    print(split,split)
elif x % 2 != 0:
    split = x // 2
    print([split, ceil(split)])
```

# Dictionary -> Use Curly Brackets

```python
#Key can be either int or string
# Value can be any object
data = {"Name":"Farah", "Age": 25, "City": "Cairo"} # Key:Value pairs
print(data)

print(type(data))

#Slicing a dictionary
print(data["Name"])  # Accessing value by key
#print(data["Address"])   # will return error if the key does not exis
print(data.get("Address"))  # Get law talabt haga msh mawgoda by default hat return None instead of error 

#Modifying a dictionary
data["Age"] = 26  # Modifying the value associated with the key "Age"
print(data)

#Addition
# method 1-> write a key with name msh mawgoda abl kdh w adelo el value el 3yzha, if the key exists it will overwrite the value
data["Addres"] = "Cairo"  # Adding a new key-value pair
print(data)

data["Addres"] = "Giza"  # Modifying the value of an existing key
print(data)

# to set a default value for a key if it does not exist
data.setdefault("Country", "Egypt")  # Adds "Country" key with value "Egypt" if it doesn't exist
data.setdefault("Country", "USA")  # Will not change the value since "Country" already exists
print(data)

tmp_dict = {1:123, 2:456}
data.update(tmp_dict)
print(data)  # Merges tmp_dict into data -> shabah el extend lel lists

```

```python
data = {"Name":"Farah", "Age": 25, "Address": "6 October", "Gender":"Female"} # Key:Value pairs

# Remove items 
data.pop("Address")  # Removes the key "Address" and its value , fl list knt badelo el index 
print(data)

data.popitem()  # Removes the last inserted key-value pair , ka2ne badelo el last index (-1)
print(data)

print(data.keys() )  # Returns a view object displaying a list of all the keys in the dictionary
print(data.values())  # Returns a view object displaying a list of all the values in the
print(data.items())  # Returns a view object displaying a list of all the key-value pairs in the dictionary
```

# Set -> Use Curly Brackets but write only values, no keywords as in dictionary
-> Set remove duplicates

```python
numbers = {12,43,32,45,12,45,65,23}
print(numbers)  # Output: {32, 65, 12, 43, 45, 23} (Duplicates are removed)
print(type(numbers))  # Output: <class 'set'>

print(numbers.difference({12, 32}))  # Returns a set with elements in numbers but not in {12, 32}
print(numbers.intersection({12, 32}))  # Returns a set with elements common to both sets
print(numbers.union({12, 32}))  # Returns a set with elements from both sets
```
