# RegEx 'Regular Expression'

```python
data = "The rain in Spain falls mainly in the plain 3 times."
print(data.find("rain"))

# find anything ending with n ? ... can't use find() -> will use regex to add pattern 

```

```python
# re should be imported to use regular expression
import re

print(re.findall('rain',data))

# pattern -> better to put r before the pattern so Python can recognize it as a raw string
# \d  will search for digits [0-9]
print(re.findall(r"\d",data))
# \D will search for any character that is NOT a digit (including spaces and punctuation)
print(re.findall(r"\D",data))
# \w will search for any alphanumeric character (letters, digits, and underscore) not spaces nor punctuation
print(re.findall(r"\w",data))
# \W will search for any character that is NOT alphanumeric (includes spaces and punctuation)
print(re.findall(r"\W",data))

```

```python
# Quantification -> ex: search for a word of 5 letters
print(re.findall(r"\w\w\w\w\w",data))
print(re.findall(r"\w\w\w\w\w ",data)) # w letters & ends with space
print(re.findall(r" \w{3} ",data)) # word of 3 letters + akherha w awelha space 34n at2kd enha word msh en hwa asele goz2 3 letters mn word atwal mn 3
print(re.findall(r" \w{3,5} ",data)) # word between 3 to 5 letters
print(re.findall(r"\w{3,}",data)) # word that is 3 or more letters
```

# quiz

```python
import re
# \s* allows for optional spaces.
# \d+ matches one or more digits.
# [\+\-\*/%] matches any one of the allowed operators.
def math_expression(s):
    pattern = r'\s*\d\s*[\+\-\*\/\%]\s*\d\s*'
    return bool(re.match(pattern,s))

print(math_expression("4 + 5"))   # True
print(math_expression("4*no"))    # False
print(math_expression(" 4 * 5 "))   # True

```
