# RegEx 'Regular Expression'

```python
data = "The rain in Spain falls mainly in the plain 3 times."
print(data.find("rain"))

# find anything ending with n ? ... can't use find() -> will use regex to add pattern 

```

```python
# re should be imported to use regular expression
import re

print(re.findall('rain',data))

# pattern -> better to put r before the pattern so Python can recognize it as a raw string
# \d  will search for digits [0-9]
print(re.findall(r"\d",data))
# \D will search for any character that is NOT a digit (including spaces and punctuation)
print(re.findall(r"\D",data))
# \w will search for any alphanumeric character (letters, digits, and underscore) not spaces nor punctuation
print(re.findall(r"\w",data))
# \W will search for any character that is NOT alphanumeric (includes spaces and punctuation)
print(re.findall(r"\W",data))

```

```python
# Quantification -> ex: search for a word of 5 letters
print(re.findall(r"\w\w\w\w\w",data))
print(re.findall(r"\w\w\w\w\w ",data)) # w letters & ends with space
print(re.findall(r" \w{3} ",data)) # word of 3 letters + akherha w awelha space 34n at2kd enha word msh en hwa asele goz2 3 letters mn word atwal mn 3
print(re.findall(r" \w{3,5} ",data)) # word between 3 to 5 letters
print(re.findall(r"\w{3,}",data)) # word that is 3 or more letters
```

# quiz

```python
import re
# \s* allows for optional spaces.
# \d+ matches one or more digits.
# [\+\-\*/%] matches any one of the allowed operators.
def math_expression(s):
    pattern = r'\s*\d\s*[\+\-\*\/\%]\s*\d\s*'
    return bool(re.match(pattern,s))

print(math_expression("4 + 5"))   # True
print(math_expression("4*no"))    # False
print(math_expression(" 4 * 5 "))   # True

```

# Task : go to Noon website (https://www.noon.com/egypt-en/all-products-eg/) and search for a specific product (iphone covers)  and get all the Products displayed with their prices in a dictionary  ex:  {"ProductName" : its_Price}

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time

# Initialize the driver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get("https://www.noon.com/egypt-en/all-products-eg/")

wait = WebDriverWait(driver, 10)
search_box = wait.until(EC.presence_of_element_located((By.ID, "search-input")))

# Search for "dior perfumes"
search_box.clear()
search_box.send_keys("dior perfumes")
search_box.send_keys(Keys.RETURN)

# Wait for results to load
time.sleep(5)

# Find all product containers using the main wrapper class 
product_containers = driver.find_elements(By.CSS_SELECTOR, "div.ProductBoxLinkHandler_linkWrapper__b0qZ9")

results = {}

for product in product_containers:

    try:
        name = product.find_element(By.CSS_SELECTOR, "h2.ProductDetailsSection_title__JorAV").text.strip()
    
        price_container = product.find_element(By.CSS_SELECTOR, "div[data-qa='plp-product-box-price']")
        
        currency = price_container.find_element(By.CLASS_NAME, "Price_currency__taKQj").text.strip()
        amount = price_container.find_element(By.CLASS_NAME, "Price_amount__2sXa7").text.strip()
        
        price = f"{currency} {amount}"
        results[name] = price
    except Exception as e:
        continue


# Print the full dictionary of product names and prices
for name,price in results.items():
    print(f"{name}:{price}")

# driver.quit()
```
