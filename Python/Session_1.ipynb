Arithmetic Operations

```python
# +,-,*,/, // "floor devision", %, **
a=10
b=3
print(a+b)  # Addition
print(a-b)  # Subtraction
print(a*b)  # Multiplication
print(a/b)  # Division
print(a//b)  # Floor Division
print(a%b)  # Modulus
print(a**b)  # Exponentiation
```

Comparision Operations

```python
#> , <, ==, !=, >=, <=
x = 5  # You can change this value to test  
y = 10  # You can change this value to test
print(x > y)   # Greater than
print(x < y)   # Less than      
print(x == y)  # Equal to
print(x != y)  # Not equal to
print(x >= y)  # Greater than or equal to
print(x <= y)  # Less than or equal to
```

Logical Operations

```python
print(x>5 and y<15)  # Logical AND
print(x>5 or y<15)   # Logical OR
print(not(x > 5))   # Logical NOT
```

String

```python
name = "Farah"
print(name)

#Slicing
print(name[0]) # First character
print(name[1:3]) #will inclucde character 1 & 2 only, 3 is not included
print(name[-1]) # Last character

print(name*3) #will print the string 3 times
print(name + " is a student") # Concatenation --> "+" here is Operator overloading
# Operator Overloading means that the same operator (like +) can behave differently depending on the data types involved.
```

```python
#Solution  -> to check whether the sum of the first two digits of a number has the same parity as the number itself
num = input("Enter a number: ")
digits_sum = int(num[0])  + int(num[1])
digits_parity = digits_sum % 2
num_parity = int(num) % 2
if digits_parity == num_parity:
    print("True")
else:
    print("False")
```
