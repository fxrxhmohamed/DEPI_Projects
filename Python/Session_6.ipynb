# Quiz

```python
class Icecream:
    def __init__(self,flavor, num_sprinkles):
        self.flavor = flavor 
        self.num_sprinkles = num_sprinkles
        
    sweet_values ={
        "Plain":0,
        "Vanilla":5,
        "ChoclateChips":5,
        "Strawberry":10,
        "Choclate":10
    }

    def sweetest_ice_cream(ice_creams):
        max_sweet = 0
        for ice_cream in ice_creams:
            value = Icecream.sweet_values[ice_cream.flavor]+ice_cream.num_sprinkles

            if value > max_sweet:
                max_sweet = value
        return max_sweet



ice1 = Icecream("Choclate",13)
ice2 = Icecream("Vanilla",10)
print(Icecream.sweetest_ice_cream([ice1, ice2]))
```

# 1) Lambda 
->an anonymous function (a function without a name) that is:
*Created using the keyword lambda
*Can take any number of arguments, but
*Can only have one expression (no multiple lines inside)
*syntax: lambda arguments: expression

```python
#Without lambda
def add(x,y):
    return x+y

print(add(4,5))
```

```python
#with lambda
add = lambda x,y : x+y
print(add(4,5))
```

```python
is_even = lambda x: True if x%2 ==0 else False
is_even(4)
```

```python
def return_numbers(n):
    for i in range(1, n):
        yield i

x = return_numbers(10000000)
print(x)

```

# All Python string methods

```python
s = "  Hello, World!  "
t = "hello"
u = "123"
v = "hello123"
w = "HELLO"
x = "hello world"
y = "Hello\nWorld"
z = "apple,banana,cherry"

# 1. capitalize
print(s.capitalize())  # '  hello, world!  '

# 2. casefold
print(s.casefold())  # '  hello, world!  '

# 3. center
print(s.center(30, '*'))  # '******  Hello, World!  ******'

# 4. count
print(s.count('l'))  # 3

# 5. encode
print(s.encode('utf-32'))  # b'  Hello, World!  '

# 6. endswith
print(s.endswith('!  '))  # True

# 7. expandtabs
tab_str = "Hello\tWorld"
print(tab_str.expandtabs(4))  # 'Hello   World'

# 8. find
print(s.find('World'))  # 9

# 9. format
print("My name is {}".format("Ahmed"))  # 'My name is Ahmed'

# 10. format_map
print("{name} is {age}".format_map({'name': 'Ahmed', 'age': 20}))  # 'Ahmed is 20'

# 11. index
print(s.index('World'))  # 9

# 12. isalnum
print(v.isalnum())  # True

# 13. isalpha
print(t.isalpha())  # True

# 14. isascii
print(s.isascii())  # True

# 15. isdecimal
print(u.isdecimal())  # True

# 16. isdigit
print(u.isdigit())  # True

# 17. isidentifier
print("my_var".isidentifier())  # True

# 18. islower
print(t.islower())  # True

# 19. isnumeric
print(u.isnumeric())  # True

# 20. isprintable
print(s.isprintable())  # True

# 21. isspace
print("   ".isspace())  # True

# 22. istitle
print("Hello World".istitle())  # True

# 23. isupper
print(w.isupper())  # True

# 24. join
print(",".join(["a", "b", "c"]))  # 'a,b,c'

# 25. ljust
print(t.ljust(10, '*'))  # 'hello*****'

# 26. lower
print(w.lower())  # 'hello'

# 27. lstrip
print(s.lstrip())  # 'Hello, World!  '

# 28. maketrans
trans = str.maketrans("H", "J")
print(s.translate(trans))  # '  Jello, World!  '

# 29. partition
print(s.partition(','))  # ('  Hello', ',', ' World!  ')

# 30. replace
print(s.replace("Hello", "Hi"))  # '  Hi, World!  '

# 31. rfind
print(s.rfind('l'))  # 11

# 32. rindex
print(s.rindex('l'))  # 11

# 33. rjust
print(t.rjust(10, '*'))  # '*****hello'

# 34. rpartition
print(s.rpartition(','))  # ('  Hello', ',', ' World!  ')

# 35. rsplit
print(z.rsplit(',', 1))  # ['apple,banana', 'cherry']

# 36. rstrip
print(s.rstrip())  # '  Hello, World!'

# 37. split
print(z.split(','))  # ['apple', 'banana', 'cherry']

# 38. splitlines
print(y.splitlines())  # ['Hello', 'World']

# 39. startswith
print(s.startswith('  He'))  # True

# 40. strip
print(s.strip())  # 'Hello, World!'

# 41. swapcase
print(s.swapcase())  # '  hELLO, wORLD!  '

# 42. title
print(x.title())  # 'Hello World'

# 43. translate
table = str.maketrans("H", "J")
print(s.translate(table))  # '  Jello, World!  '

# 44. upper
print(t.upper())  # 'HELLO'

# 45. zfill
print(u.zfill(6))  # '000123
```
