# Quiz

```python
class Icecream:
    def __init__(self,flavor, num_sprinkles):
        self.flavor = flavor 
        self.num_sprinkles = num_sprinkles
        
    sweet_values ={
        "Plain":0,
        "Vanilla":5,
        "ChoclateChips":5,
        "Strawberry":10,
        "Choclate":10
    }

    def sweetest_ice_cream(ice_creams):
        max_sweet = 0
        for ice_cream in ice_creams:
            value = Icecream.sweet_values[ice_cream.flavor]+ice_cream.num_sprinkles

            if value > max_sweet:
                max_sweet = value
        return max_sweet



ice1 = Icecream("Choclate",13)
ice2 = Icecream("Vanilla",10)
print(Icecream.sweetest_ice_cream([ice1, ice2]))
```

# 1) Lambda 
->an anonymous function (a function without a name) that is:
*Created using the keyword lambda
*Can take any number of arguments, but
*Can only have one expression (no multiple lines inside)
*syntax: lambda arguments: expression

```python
#Without lambda
def add(x,y):
    return x+y

print(add(4,5))
```

```python
#with lambda
add = lambda x,y : x+y
print(add(4,5))
```

```python
is_even = lambda x: True if x%2 ==0 else False
is_even(4)
```

# List statement

```python
#Normal for loop
#  its problem with lambda as it is not in one line
numbers = []
for i in range(1,11):
    numbers.append(i)
print(numbers)
```

```python
#  solution to be able to put for loop in lambda
numbers = [i for i in range(1,11)]
print(numbers)
```

```python
# Using lambda to use for loop
hello = lambda n: [i for i in range(n)]
hello(3)
```

# Generataors

```python
def return_numbers(n):
    for i in range(1, n):
        yield i

x = return_numbers(10000000)
print(x)

```

# Python func tools

```python
# reduce -> will reduce the elements in the list tell one is left and implement the expression u will define on this elements in the lambda 
from functools import reduce
nums = [12,54,73,1,8,4,66,68,45,21]
my_reduce = reduce(lambda x,y:x+y,nums)
print(my_reduce)  # 352
```

```python
# zip -> combine more than element , list in one place
num = [1,2,4,6]
name = ["farah","moh","abdallah","yahya"]
print(list(zip(num,name)))  #[(1, 'farah'), (2, 'moh'), (4, 'abdallah'), (6, 'yahya')]
```
