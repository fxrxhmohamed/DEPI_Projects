```python
# connect to the datatbase 
import mysql.connector
conn = mysql.connector.connect(
    user='root',
    database='depi3'
)
cursor = conn.cursor()
```

```python
# this shows all the databases i have in mysql
cursor.execute("SHOW DATABASES")
print(cursor.fetchall())
```

```python
# this shows all the tables i have 
cursor.execute("SHOW TABLES")
print(cursor.fetchall())
```

```python
cursor.execute("INSERT INTO employee (id, name, salary) VALUES (1,'John Doe', 50000)")
```

```python
data = [
    (2, 'john', 20, 30000),
    (3, 'farah', 22, 60000),
    (4, 'farida', 10, 50000)
]
cursor.executemany("INSERT INTO Employee (id, name, age, salary) VALUES (%s, %s, %s, %s)", data)
```

```python
cursor.execute("SELECT * FROM employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Comparison Operators
cursor.execute("SELECT * FROM employee WHERE salary <> 60000 ")
records = cursor.fetchall()
print(records)
```

```python
# select using IN
cursor.execute("SELECT * FROM employee WHERE salary IN (50000,6000)")
records = cursor.fetchall()
print(records)
```

```python
# select using ALL
cursor.execute("SELECT salary FROM employee WHERE salary = ALL (SELECT salary FROM employee WHERE name = 'farah')")
records = cursor.fetchall()
print(records)
```

```python
# select using LIKE
cursor.execute("SELECT * FROM employee WHERE name LIKE 'j%'")
records = cursor.fetchall()
print(records)
```

```python
# select using LIKE -> similar to the regEx
# _a% means: the second character in the string must be a, no matter what the first character is.
cursor.execute("SELECT * FROM employee WHERE name LIKE '_a%'") 
records = cursor.fetchall()
print(records)
```
