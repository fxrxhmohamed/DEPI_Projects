# JOIN

```python
# Create tables for JOIN demonstration

# Create departments table

cursor.execute("""
CREATE TABLE IF NOT EXISTS departments (
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(50),
    location VARCHAR(50)
)
""")

# Create employees table with department reference
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(50),
    salary DECIMAL(10,2),
    dept_id INT,
    hire_date DATE
)
""")

# Create projects table
cursor.execute("""
CREATE TABLE IF NOT EXISTS projects (
    project_id INT PRIMARY KEY,
    project_name VARCHAR(100),
    budget DECIMAL(12,2),
    dept_id INT
)
""")

# Create employee_projects table (many-to-many relationship)
cursor.execute("""
CREATE TABLE IF NOT EXISTS employee_projects (
    emp_id INT,
    project_id INT,
    role VARCHAR(50),
    PRIMARY KEY (emp_id, project_id)
)
""")
```

# Outer Join
- Left Join -> return all left table rows even if no match in the right we law fe common mabenhom haygebo 3adi
- Right Join -> return all right table rows even if no match in the left we law fe common mabenhom haygebo 3adi
- Full Join -> by3ml union el mn hena w hena bs will remove dublicates

```python
cursor.execute("""
               SELECT e.emp_id, e.emp_name, ep.role 
               FROM employees e, employee_projects ep
               WHERE e.emp_id = ep.emp_id
""")
#list column names
columns = [column[0] for column in cursor.description]
print(columns)
records = cursor.fetchall()
for record in records :
    print(record)
```

```python
# Join between 3 tables -> employees, employee_project, projects
cursor.execute("""
               SELECT e.emp_id, e.emp_name, ep.role 
               FROM employees e RIGHT OUTER JOIN employee_projects ep
               ON e.emp_id = ep.emp_id
               RIGHT OUTER JOIN projects p
               ON ep.project_id = p.project_id
               ORDER BY e.emp_id ASC
""")
#list column names
columns = [column[0] for column in cursor.description]
print(columns)
records = cursor.fetchall()
for record in records :
    print(record)
```

```python
# make further filteration for specific project name
cursor.execute("""
               SELECT e.emp_id, e.emp_name, ep.role 
               FROM employees e RIGHT OUTER JOIN employee_projects ep
               ON e.emp_id = ep.emp_id
               RIGHT OUTER JOIN projects p
               ON ep.project_id = p.project_id
               WHERE p.project_name = 'Website Redesign'
               ORDER BY e.emp_id ASC
""")
#list column names
columns = [column[0] for column in cursor.description]
print(columns)
records = cursor.fetchall()
for record in records :
    print(record)
```
