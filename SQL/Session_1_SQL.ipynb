# DML
-INSERT
-UPDATE
-DELETE

```python

# Insert 1 row
cursor.execute("INSERT INTO Employee (id, name, age) VALUES (1, 'Farah Moh', 21)")
cursor.execute("SELECT * FROM Employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Insert multiple data
data = [
    (2,'Salma', 22),
    (3,'Abd', 23),
    (4, 'Doaa',25)
]

# %s is a placeholder for a value (even for strings and numbers in MySQL
cursor.executemany("INSERT INTO Employee (id, name, age) VALUES (%s, %s, %s)", data) 

cursor.execute("SELECT * FROM Employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Update some exisitng records

cursor.execute("UPDATE Employee SET salary = 5000 WHERE id = 1")

cursor.execute("SELECT * FROM Employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Update some exisitng records

cursor.execute("UPDATE Employee SET salary = 3000 WHERE age > 21")

cursor.execute("SELECT * FROM Employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# DELETE 

cursor.execute("DELETE FROM Employee WHERE id = 1")

cursor.execute("SELECT * FROM Employee")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
